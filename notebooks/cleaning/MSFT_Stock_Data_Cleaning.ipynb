```python
import yfinance as yf
import pandas as pd
```

Here I downloaded the MSFT stock data for the past 5 years. I used head() and tail() just to have a sanity check to ensure all data was in ascending order and loaded in properly.

> Add blockquote



```python
msft = yf.download("MSFT", period="5y", auto_adjust=True)
msft.head()
```

```python
msft.tail()
```

Using shape was useful because it showed me how many rows and columns were in the dataset, which helped me confirm the dataset size and understand how much data I was working with.

```python
msft.shape
```

Info just provided me with each column and its datatype. This was also loaded into understand the specific data types I was dealing with which all looked good.

```python
msft.info()
```

Checking for null values was important to rule out any missing values which needed to be removed. The sum just provides how many missing values there were.

```python
msft.isnull().sum()
```

Checking for duplicates was important since we do not want repeating rows.

```python
msft.index.duplicated().sum()
```

The describe() function is useful because it provides a statistical summary of the dataset. This helps identify unusual values or patterns, such as unexpectedly high or low values, that may need to be examined more closely.

```python
msft.describe()
```

```python
msft.reset_index(inplace=True)
msft.columns = msft.columns.droplevel(1)
msft.head()
```

The following lines of code was just to check whether any of the columns had any negative values which shouldn't be there. It was done as a sanity check.

```python
print((msft["Close"] <= 0).sum())
print((msft["High"] <= 0).sum())
print((msft["Low"] <= 0).sum())
print((msft["Open"] <= 0).sum())
print((msft["Volume"] < 0).sum())
```

```python
(msft["Volume"] < 0).sum()
```

I checked whether any Highs may be less than the lows. Even though it is highly unlikely for that to ever happen it was just an important sanity check.

```python
(msft["High"] < msft["Low"]).sum()
```

```python
msft.columns = msft.columns.get_level_values(0)
msft.head()
```

Overall, the data itself was clean there wasn't anything that reall had to be done here. All data was saved to a csv which can be found in the folder within this colab.

```python
msft.to_csv("../Data/msft.csv", index = False)
```
