```python
'''It reads daily close, high, low, open, and volume form the trading days in the five years period.'''
import yfinance as yf
import pandas as pd

#download five year data from Bloom Energy, adjusted for split-shares and dividends
be = yf.download("BE", period="5y", auto_adjust=True)
print(be.shape)
be.head()
```

```python
# Bloom Energy has no dividend; it *never* split-share.
ticker = yf.Ticker("BE")
print(ticker.splits)
print(ticker.dividends)
```

Prices columns should only have two decimal points since no split-share took place.

```python
# No missing value, non invalid values, no duplication
print(be.isnull().sum())
print((be <= 0).sum())
print(be.index.duplicated().sum())
```

```python
be.reset_index(inplace=True)
be.columns = be.columns.droplevel(1)
be.head()
```

```python
# Save to CSV
be.to_csv("../Data/BE.csv", index = False)
```
