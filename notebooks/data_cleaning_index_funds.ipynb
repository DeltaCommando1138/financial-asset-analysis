```python
import numpy as np
import pandas as pd
import yfinance as yf
import seaborn as sns
```

# S&P 500 SPY ETF

```python
df = yf.download("SPY", period = "5y", auto_adjust = True)
```

```python
df.head()
```

```python
df.tail()
```

```python
df.reset_index(inplace=True)
df.head()
```

```python
df.columns = df.columns.droplevel(1)
df.head()
```

```python
df.shape
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()
```

```python
print(df.index.duplicated().sum())
```

```python
df.info()
```

```python
df.drop(columns = ['Date']).describe()
```

```python
sns.histplot(data = df, x = 'Close')
```

Distribution of SPY Close date prices appears to be right-skewed, meaning that most prices tend to be on the lower side.

```python
sns.histplot(data = df, x = 'Volume')
```

Right-skewed distribution of SPY ETF Volume show that trading tends to be on the low-moderate frequencies.

```python
df.head()
```

```python
df.to_csv('../Data/spy.csv', index = False)
```

# S&P 500 Vanguard ETF

```python
van_df = yf.download("VOO", period = "5y", auto_adjust = True)
```

```python
van_df.head()
```

```python
van_df.tail()
```

```python
van_df.reset_index(inplace=True)
van_df.head()
```

```python
van_df.columns = van_df.columns.droplevel(1)
van_df.head()
```

```python
print(van_df.isnull().sum())
```

```python
print(van_df.index.duplicated().sum())
```

```python
van_df = van_df.dropna()
```

```python
van_df.shape
```

```python
sns.histplot(data = van_df, x = 'Close')
```

Distribution of Vanguard index appears right-skewed, and range appears to be slightly lower compared to SPY's distribution.

```python
sns.histplot(data = van_df, x = 'Volume')
```

Trade volumes for Vanguard appear to be heavily right-skewed, indicating that trading frequencies also tend to be on the low-moderate side within our data.

```python
van_df.to_csv("../Data/vanguard.csv", index = False)
```
